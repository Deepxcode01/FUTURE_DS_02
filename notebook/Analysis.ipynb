## 1️⃣ Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.figsize'] = (10, 5)
plt.rcParams['font.family'] = 'DejaVu Sans'
print('✅ Libraries loaded successfully')
```

## 2️⃣ Load Dataset

```python
df = pd.read_csv('../dataset/telecom_churn_data.csv')
print(f'Shape: {df.shape}')
df.head()
```

```python
print('📋 Dataset Info:')
df.info()
print('\n📊 Missing Values:')
print(df.isnull().sum())
```

```python
print('📈 Basic Statistics:')
df.describe().round(2)
```

## 3️⃣ Data Cleaning

```python
# Check and fix data types
df['JoinDate'] = pd.to_datetime(df['JoinDate'])
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df.dropna(subset=['TotalCharges'], inplace=True)

# Derived columns
df['Churn_Label'] = df['Churn'].map({1: 'Churned', 0: 'Active'})
df['Tenure_Group'] = pd.cut(df['Tenure_Months'],
    bins=[0,6,12,24,36,72],
    labels=['0-6 months','7-12 months','13-24 months','25-36 months','36+ months'])
df['Charge_Group'] = pd.cut(df['MonthlyCharges'],
    bins=[0,30,50,70,90,120],
    labels=['₹0-30','₹31-50','₹51-70','₹71-90','₹90+'])

print(f'✅ Cleaned dataset: {df.shape[0]} rows')
print(f'✅ Overall Churn Rate: {df["Churn"].mean()*100:.1f}%')
```

## 4️⃣ KPI Summary

```python
total = len(df)
churned = df['Churn'].sum()
active = total - churned
churn_rate = churned/total*100
avg_tenure = df['Tenure_Months'].mean()
avg_monthly = df['MonthlyCharges'].mean()
revenue_lost = df[df['Churn']==1]['TotalCharges'].sum()

print('='*45)
print(f'  Total Customers     : {total:,}')
print(f'  Active Customers    : {active:,}')
print(f'  Churned Customers   : {churned:,}')
print(f'  Churn Rate          : {churn_rate:.1f}%')
print(f'  Avg Tenure          : {avg_tenure:.1f} months')
print(f'  Avg Monthly Charge  : ₹{avg_monthly:.2f}')
print(f'  Revenue at Risk     : ₹{revenue_lost:,.0f}')
print('='*45)
```

## 5️⃣ Univariate Analysis

```python
fig, axes = plt.subplots(1, 2, figsize=(12,5))

# Churn distribution
colors = ['#2ecc71','#e74c3c']
df['Churn_Label'].value_counts().plot(kind='bar', ax=axes[0], color=colors, edgecolor='white', rot=0)
axes[0].set_title('Overall Churn Distribution', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Number of Customers')
for p in axes[0].patches:
    axes[0].annotate(f'{int(p.get_height())}', (p.get_x()+p.get_width()/2., p.get_height()), ha='center', va='bottom', fontweight='bold')

# Contract type
df['Contract'].value_counts().plot(kind='bar', ax=axes[1], color=['#3498db','#9b59b6','#f39c12'], edgecolor='white', rot=15)
axes[1].set_title('Contract Type Distribution', fontsize=13, fontweight='bold')
axes[1].set_ylabel('Number of Customers')
for p in axes[1].patches:
    axes[1].annotate(f'{int(p.get_height())}', (p.get_x()+p.get_width()/2., p.get_height()), ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.savefig('../images/01_churn_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12,5))

# Monthly charges
df['MonthlyCharges'].plot(kind='hist', bins=30, ax=axes[0], color='#3498db', edgecolor='white', alpha=0.8)
axes[0].set_title('Monthly Charges Distribution', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Monthly Charges (₹)')
axes[0].axvline(df['MonthlyCharges'].mean(), color='red', linestyle='--', label=f'Mean: ₹{df["MonthlyCharges"].mean():.1f}')
axes[0].legend()

# Tenure
df['Tenure_Months'].plot(kind='hist', bins=30, ax=axes[1], color='#9b59b6', edgecolor='white', alpha=0.8)
axes[1].set_title('Tenure Distribution', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Tenure (Months)')
axes[1].axvline(df['Tenure_Months'].mean(), color='red', linestyle='--', label=f'Mean: {df["Tenure_Months"].mean():.1f}m')
axes[1].legend()

plt.tight_layout()
plt.savefig('../images/02_charges_tenure_dist.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 6️⃣ Bivariate Analysis — Churn Drivers

```python
fig, axes = plt.subplots(1, 2, figsize=(13,5))

# Churn by Contract
ct = df.groupby('Contract')['Churn'].mean().mul(100).round(1).sort_values(ascending=False)
bars = ct.plot(kind='bar', ax=axes[0], color=['#e74c3c','#f39c12','#2ecc71'], edgecolor='white', rot=15)
axes[0].set_title('Churn Rate by Contract Type', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Churn Rate (%)')
for p in axes[0].patches:
    axes[0].annotate(f'{p.get_height():.1f}%', (p.get_x()+p.get_width()/2., p.get_height()), ha='center', va='bottom', fontweight='bold')

# Churn by Payment Method
pm = df.groupby('PaymentMethod')['Churn'].mean().mul(100).round(1).sort_values(ascending=False)
pm.plot(kind='bar', ax=axes[1], color='#e67e22', edgecolor='white', rot=20)
axes[1].set_title('Churn Rate by Payment Method', fontsize=13, fontweight='bold')
axes[1].set_ylabel('Churn Rate (%)')
for p in axes[1].patches:
    axes[1].annotate(f'{p.get_height():.1f}%', (p.get_x()+p.get_width()/2., p.get_height()), ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.savefig('../images/03_churn_contract_payment.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13,5))

# Churn by Tenure Group
tg = df.groupby('Tenure_Group', observed=True)['Churn'].mean().mul(100).round(1)
tg.plot(kind='bar', ax=axes[0], color='#1abc9c', edgecolor='white', rot=20)
axes[0].set_title('Churn Rate by Tenure Group', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Churn Rate (%)')
for p in axes[0].patches:
    axes[0].annotate(f'{p.get_height():.1f}%', (p.get_x()+p.get_width()/2., p.get_height()), ha='center', va='bottom', fontweight='bold')

# Churn by Internet Service
is_ = df.groupby('InternetService')['Churn'].mean().mul(100).round(1).sort_values(ascending=False)
is_.plot(kind='bar', ax=axes[1], color=['#e74c3c','#3498db','#2ecc71'], edgecolor='white', rot=10)
axes[1].set_title('Churn Rate by Internet Service', fontsize=13, fontweight='bold')
axes[1].set_ylabel('Churn Rate (%)')
for p in axes[1].patches:
    axes[1].annotate(f'{p.get_height():.1f}%', (p.get_x()+p.get_width()/2., p.get_height()), ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.savefig('../images/04_churn_tenure_internet.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Boxplot: Monthly Charges vs Churn
fig, axes = plt.subplots(1, 2, figsize=(13,5))

sns.boxplot(data=df, x='Churn_Label', y='MonthlyCharges', palette={'Active':'#2ecc71','Churned':'#e74c3c'}, ax=axes[0])
axes[0].set_title('Monthly Charges vs Churn', fontsize=13, fontweight='bold')
axes[0].set_xlabel('')

sns.boxplot(data=df, x='Churn_Label', y='Tenure_Months', palette={'Active':'#3498db','Churned':'#e74c3c'}, ax=axes[1])
axes[1].set_title('Tenure vs Churn', fontsize=13, fontweight='bold')
axes[1].set_xlabel('')

plt.tight_layout()
plt.savefig('../images/05_boxplots_charges_tenure.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 7️⃣ Churn Reasons Analysis

```python
reasons = df[df['Churn']==1]['ChurnReason'].value_counts()

fig, axes = plt.subplots(1,2, figsize=(13,5))
reasons.plot(kind='barh', ax=axes[0], color='#e74c3c', edgecolor='white')
axes[0].set_title('Top Churn Reasons', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Number of Customers')
for p in axes[0].patches:
    axes[0].annotate(f'{int(p.get_width())}', (p.get_width(), p.get_y()+p.get_height()/2.), ha='left', va='center', fontweight='bold')

axes[1].pie(reasons, labels=reasons.index, autopct='%1.1f%%', startangle=140,
    colors=['#e74c3c','#e67e22','#f39c12','#3498db','#9b59b6','#1abc9c'])
axes[1].set_title('Churn Reasons (Pie)', fontsize=13, fontweight='bold')

plt.tight_layout()
plt.savefig('../images/06_churn_reasons.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 8️⃣ Cohort Retention Analysis

```python
cohort = df.groupby('CohortMonth').agg(
    Total=('Churn','count'),
    Churned=('Churn','sum')
).reset_index()
cohort['RetentionRate'] = ((cohort['Total']-cohort['Churned'])/cohort['Total']*100).round(1)
cohort['ChurnRate'] = (cohort['Churned']/cohort['Total']*100).round(1)
cohort = cohort.sort_values('CohortMonth')

fig, ax = plt.subplots(figsize=(14,5))
ax.plot(cohort['CohortMonth'], cohort['RetentionRate'], marker='o', color='#2ecc71', linewidth=2.5, markersize=6, label='Retention Rate %')
ax.plot(cohort['CohortMonth'], cohort['ChurnRate'], marker='s', color='#e74c3c', linewidth=2.5, markersize=6, label='Churn Rate %')
ax.axhline(cohort['RetentionRate'].mean(), color='#2ecc71', linestyle='--', alpha=0.5, label=f'Avg Retention: {cohort["RetentionRate"].mean():.1f}%')
ax.set_title('Monthly Cohort — Retention vs Churn Rate Over Time', fontsize=13, fontweight='bold')
ax.set_xlabel('Join Month')
ax.set_ylabel('Rate (%)')
ax.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('../images/07_cohort_retention.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 9️⃣ Correlation Heatmap

```python
num_cols = ['Age','Tenure_Months','MonthlyCharges','TotalCharges','NumProducts','SeniorCitizen','Churn']
corr = df[num_cols].corr().round(2)

plt.figure(figsize=(9,7))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdYlGn', center=0,
    linewidths=0.5, square=True, annot_kws={'size':10})
plt.title('Correlation Heatmap', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig('../images/08_correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
```
